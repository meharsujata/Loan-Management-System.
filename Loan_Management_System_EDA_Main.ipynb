## 1. Load the dataset

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

DATA_PATH = "lending_club_loan_two.csv"

df = pd.read_csv(DATA_PATH)

print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

df.head()
```

## 3. Clean the data

Create a copy so the original dataset remains unchanged. Fill missing values in the three text columns below with clear labels. Other missing values are left unchanged for this analysis.

```python
df_clean = df.copy()

df_clean['emp_title'] = df_clean['emp_title'].fillna('Unknown')
df_clean['emp_length'] = df_clean['emp_length'].fillna('Unknown')
df_clean['title'] = df_clean['title'].fillna('Not Provided')

df_clean.isnull().sum()
```

```python
df_clean.duplicated().sum()
```

## 4. Exploratory Data Analysis

The following 12 analyses focus on loan outcomes, loan characteristics and borrower information that may help shape LMS monitoring and reporting requirements.

### 1. Loan status distribution

Shows how many loans were Fully Paid or Charged Off, and the percentage represented by each outcome.

```python
df_clean['loan_status'].value_counts()
```

```python
loan_status_pct = df_clean['loan_status'].value_counts(normalize=True) * 100
loan_status_pct.round(2)
```

### 2. Loan amount by loan status

Compares the number, average amount and total amount of loans for each historical outcome.

```python
df_clean.groupby('loan_status')['loan_amnt'].agg(['count', 'mean', 'sum']).round(2)
```

### 3. Loan purpose distribution

Identifies the most common reasons borrowers gave for taking loans.

```python
df_clean['purpose'].value_counts()
```

### 4. Loan grade distribution

Shows how the loan portfolio is distributed across LendingClub grades.

```python
df_clean['grade'].value_counts().sort_index()
```

### 5. Loan grade and loan status

Compares the percentage of Fully Paid and Charged Off loans within each grade.

```python
grade_status = pd.crosstab(df_clean['grade'], df_clean['loan_status'], normalize='index') * 100
grade_status.round(2)
```

### 6. Average interest rate by grade

Compares average interest rates across loan grades.

```python
df_clean.groupby('grade')['int_rate'].mean().round(2)
```

### 7. Employment length distribution

Shows the number of records in each employment-length category, including unknown values.

```python
df_clean['emp_length'].value_counts()
```

### 8. Average loan amount by purpose

Compares the average loan amount across different loan purposes.

```python
df_clean.groupby('purpose')['loan_amnt'].mean().sort_values(ascending=False).round(2)
```

### 9. Average loan amount by employment length

Compares average loan amounts across employment-length categories.

```python
df_clean.groupby('emp_length')['loan_amnt'].mean().sort_values(ascending=False).round(2)
```

### 10. DTI and loan status

Groups debt-to-income (DTI) values and compares loan outcomes within each DTI group.

```python
df_clean['DTI_Group'] = pd.cut(
    df_clean['dti'],
    bins=[-float('inf'), 10, 20, 30, 40, float('inf')],
    labels=['Below 10', '10–20', '20–30', '30–40', 'Above 40']
)

dti_status = pd.crosstab(
    df_clean['DTI_Group'],
    df_clean['loan_status'],
    normalize='index'
) * 100

dti_status.round(2)
```

### 11. Verification status and loan status

Compares loan outcomes across the dataset's verification-status categories. This is an association, not proof that verification causes an outcome.

```python
verification_status = pd.crosstab(
    df_clean['verification_status'],
    df_clean['loan_status'],
    normalize='index'
) * 100

verification_status.round(2)
```

### 12. Loan term and loan status

Compares the percentage of Fully Paid and Charged Off loans for each loan term.

```python
term_status = pd.crosstab(
    df_clean['term'],
    df_clean['loan_status'],
    normalize='index'
) * 100

term_status.round(2)
```

## 5. Key business observations

- The dataset contains two loan outcomes: **Fully Paid** and **Charged Off**.
- Charged-off rates vary across loan grades.
- Average interest rates differ across loan grades.
- Higher DTI groups show a higher charged-off share in this dataset.
- Charged-off shares differ across verification-status categories and loan terms.
- Debt consolidation and credit card are the most common loan purposes.

These are historical patterns in this dataset. They should not be interpreted as proof of causation or as a credit-approval policy.

## 6. Suggested LMS capabilities

Based on the analysis, a proposed LMS could include:

- **Loan records:** store loan amount, purpose, grade, interest rate and term.
- **Applicant information:** record employment length, annual income and DTI.
- **Verification tracking:** record verification status and supporting review details.
- **Portfolio monitoring:** report loan status and charged-off rates by grade, term and other categories.
- **Management reporting:** provide filters and summaries for monitoring the loan portfolio.

These are proposed system capabilities. The dataset does not confirm that these workflows currently exist.

## 7. Limitations and conclusion

This analysis uses historical LendingClub loan records. It does not contain full application-stage history, document-review timestamps, staff assignments, actual instalment payment records or collections notes. Therefore, it cannot measure application turnaround time, missed EMI events or collection performance.

The analysis identifies patterns associated with historical loan outcomes and provides a starting point for defining LMS data fields and reporting requirements. Further requirements would need validation with business stakeholders.
